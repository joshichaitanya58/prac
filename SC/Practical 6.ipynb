### Practical 6 – Fuzzy Membership Functions

```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz
```

```python
# Triangular Membership Function

x = np.arange(0, 11, 1)
y = fuzz.trimf(x, [3, 6, 10])

plt.plot(x, y)
plt.xlabel("Triangular Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("Triangular Membership Function")
plt.show()
```

```python
# Trapezoidal Membership Function

x = np.arange(0, 11, 1)
y = fuzz.trapmf(x, [1, 4, 6, 9])

plt.plot(x, y)
plt.xlabel("Trapezoidal Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("Trapezoidal Membership Function")
plt.show()
```

```python
# Gaussian Membership Function

x = np.arange(0, 10.1, 0.1)
y = fuzz.gaussmf(x, 4, 1)

plt.plot(x, y)
plt.xlabel("Gaussian Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("Gaussian Membership Function")
plt.show()
```

```python
# Gaussian 2 Membership Function

x = np.arange(0, 10.1, 0.1)

y = fuzz.gauss2mf(x, 1, 4, 5, 7)

plt.plot(x, y)
plt.xlabel("Gaussian 2 Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("Gaussian 2 Membership Function")
plt.show()
```

```python
# Bell Membership Function

x = np.arange(0, 10.1, 0.1)
y = fuzz.gbellmf(x, 1, 4, 7)

plt.plot(x, y)
plt.xlabel("Bell Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("Bell Membership Function")
plt.show()
```

```python
# Sigmoidal Membership Function

x = np.arange(0, 10.1, 0.1)
y = fuzz.sigmf(x, 4, 1)

plt.plot(x, y)
plt.xlabel("Sigmoidal Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("Sigmoidal Membership Function")
plt.show()
```

```python
# Sigmoidal 2 Membership Function

x = np.arange(0, 10.1, 0.1)
y = fuzz.smf(x, 1, 4)

plt.plot(x, y)
plt.xlabel("Sigmoidal 2 Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("Sigmoidal 2 Membership Function")
plt.show()
```

```python
# PI Membership Function

x = np.arange(0, 10.1, 0.1)
y = fuzz.pimf(x, 1, 4, 5, 7)

plt.plot(x, y)
plt.xlabel("PI Membership Function")
plt.ylabel("Fuzziness")
plt.ylim(-0.05, 1.05)
plt.title("PI Membership Function")
plt.show()
```
