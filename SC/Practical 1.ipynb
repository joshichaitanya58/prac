### Implement simple logic network using MP neuron model

```python
# AND Gate using M-P Neuron Model

x1 = [0, 0, 1, 1]
x2 = [0, 1, 0, 1]

w1 = [1, 1, 1, 1]
w2 = [1, 1, 1, 1]

t = 2

print("x1   x2   w1    w2    t     O")

for i in range(len(x1)):
    net = x1[i] * w1[i] + x2[i] * w2[i]

    if net >= t:
        output = 1
    else:
        output = 0

    print(x1[i], "   ", x2[i], "   ", w1[i], "   ",
          w2[i], "    ", t, "   ", output)
```

```python
# OR Gate using M-P Neuron Model

x1 = [0, 0, 1, 1]
x2 = [0, 1, 0, 1]

w1 = [1, 1, 1, 1]
w2 = [1, 1, 1, 1]

t = 1

print("x1   x2   w1   w2   t    O")

for i in range(len(x1)):
    net = x1[i] * w1[i] + x2[i] * w2[i]

    if net >= t:
        output = 1
    else:
        output = 0

    print(x1[i], "   ", x2[i], "   ", w1[i], "   ",
          w2[i], "   ", t, "   ", output)
```

```python
# XOR Gate using M-P Neuron Model

x1 = [0, 0, 1, 1]
x2 = [0, 1, 0, 1]

w1 = [1, -1, -1, 1]
w2 = [1, 1, -1, -1]

t1 = 0
t2 = 1

print(f"{'x1':<4} {'x2':<4} {'w1':<4} {'w2':<4} {'t1':<4} {'t2':<4} {'O':<2}")

for i in range(len(x1)):

    net1 = x1[i] * w1[i] + x2[i] * w2[i]

    o1 = 1 if net1 >= t1 else 0

    net2 = o1 * w1[i] + (1 - o1) * w2[i]

    output = 1 if net2 >= t2 else 0

    print(f"{x1[i]:<4} {x2[i]:<4} {w1[i]:<4} {w2[i]:<4} {t1:<4} {t2:<4} {output:<2}")
```

```python
# NOR Gate using M-P Neuron Model

x1 = [0, 0, 1, 1]
x2 = [0, 1, 0, 1]

w1 = [-1, -1, -1, -1]
w2 = [-1, -1, -1, -1]

t = 0

print("x1   x2   w1   w2   t    O")

for i in range(len(x1)):
    net = x1[i] * w1[i] + x2[i] * w2[i]

    if net >= t:
        output = 1
    else:
        output = 0

    print(x1[i], "   ", x2[i], "   ", w1[i], "   ",
          w2[i], "   ", t, "   ", output)
```
