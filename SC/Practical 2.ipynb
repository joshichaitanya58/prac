### Linear Regression using Single Neuron Model

```python


import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate random linear data
np.random.seed(101)

x = np.linspace(0, 25, 100)
y = np.linspace(0, 25, 100)

# Add noise to the random linear data
x += np.random.uniform(-4, 4, 100)
y += np.random.uniform(-4, 4, 100)

n = len(x)

# Step 2: Plot Training Data
plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Training Data")
plt.show()

# Step 3: Initialize weight and bias
W = np.random.randn()
b = np.random.randn()

# Learning rate
learning_rate = 0.001

# Training epochs
training_epochs = 1000

# Step 4: Gradient Descent
for epoch in range(training_epochs):

    # Hypothesis
    y_pred = W * x + b

    # Mean Squared Error Cost Function
    cost = np.sum((y_pred - y) ** 2) / (2 * n)

    # Gradients
    dW = np.sum((y_pred - y) * x) / n
    db = np.sum(y_pred - y) / n

    # Update weight and bias
    W = W - learning_rate * dW
    b = b - learning_rate * db

    # Display result every 50 epochs
    if (epoch + 1) % 50 == 0:
        print(
            "Epoch:", epoch + 1,
            ": cost =", cost,
            "W =", W,
            "b =", b
        )

# Step 5: Final values
training_cost = cost
weight = W
bias = b

# Step 6: Calculate predictions
predictions = weight * x + bias

print("\nTraining cost =", training_cost)
print("Weight =", weight)
print("Bias =", bias)

# Step 7: Plot Linear Regression Result
plt.plot(x, y, 'ro', label="Original data")
plt.plot(x, predictions, label="Fitted line")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression Result")
plt.legend()
plt.show()

# Step 8: Conclusion
print("\nConclusion: Linear Regression using single neural network/model is implemented successfully.")
```
